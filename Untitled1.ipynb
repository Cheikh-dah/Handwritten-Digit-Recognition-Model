```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

print("Environment ready!")
```

```python
from sklearn.datasets import fetch_openml

X, y = fetch_openml(
    "mnist_784",
    version=1,
    return_X_y=True,
    as_frame=False
)

print("Data loaded successfully!")
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print(X[0])
print("Label:", y[0])
```

```python
X = X.astype(np.float32) / 255.0

print("Minimum:", X.min())
print("Maximum:", X.max())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=10000,
    random_state=42,
    stratify=y
)

print("Training images:", X_train.shape)
print("Testing images:", X_test.shape)
print("Training labels:", y_train.shape)
print("Testing labels:", y_test.shape)
```

```python
# Network architecture
input_size = 784
hidden_size = 128
output_size = 10

# Initialize random weights
rng = np.random.default_rng(42)

W1 = rng.normal(
    0, np.sqrt(2 / input_size),
    (input_size, hidden_size)
)

b1 = np.zeros((1, hidden_size))

W2 = rng.normal(
    0, np.sqrt(2 / hidden_size),
    (hidden_size, output_size)
)

b2 = np.zeros((1, output_size))

print("W1:", W1.shape)
print("b1:", b1.shape)
print("W2:", W2.shape)
print("b2:", b2.shape)
```

```python
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return (x > 0).astype(float)

def softmax(x):
    exp_x = np.exp(x - np.max(x, axis=1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=1, keepdims=True)
```

```python
values = np.array([-3, -1, 0, 2, 5])

print("Original:", values)
print("After ReLU:", relu(values))
```

```python
def forward(X):
    # Input → Hidden layer
    Z1 = X @ W1 + b1
    A1 = relu(Z1)

    # Hidden → Output layer
    Z2 = A1 @ W2 + b2
    A2 = softmax(Z2)

    return Z1, A1, A2
```

```python
Z1, A1, A2 = forward(X_train[:1])

print("Predicted digit:", np.argmax(A2))
print("Actual digit:", y_train[0])
print("Probabilities:", A2)
```

```python
def compute_loss(predictions, labels):
    m = len(labels)

    correct_probs = predictions[np.arange(m), labels.astype(int)]

    loss = -np.mean(np.log(np.clip(correct_probs, 1e-12, 1.0)))

    return loss


loss = compute_loss(A2, y_train[:1])

print("Loss:", loss)
```

```python
# Forward propagation on testing data
Z1, A1, A2 = forward(X_test)

# Select the digit with the highest probability
predictions = np.argmax(A2, axis=1)

# Compare predictions with actual labels
accuracy = np.mean(predictions == y_test.astype(int))

print(f"Test Accuracy: {accuracy * 100:.2f}%")
```
